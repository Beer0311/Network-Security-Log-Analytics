```python
import pandas as pd


path = '/Users/beerbhadarsingh/Documents/log analysis/cybersecurity_attacks.csv'

df = pd.read_csv(path)

# This will show you the data!
print("Data ingestion successful!")
df.head()
```

```python
# 1. Convert Timestamp to proper format
df['Timestamp'] = pd.to_datetime(df['Timestamp'])

# 2. Create a 'Threat_Score' 
# High severity packets or packets over 1200 bytes get a higher score.
def calculate_score(row):
    score = 0
    if row['Severity Level'] == 'High': score += 5
    elif row['Severity Level'] == 'Medium': score += 3
    
    if row['Packet Length'] > 1200: score += 5
    return score

df['Threat_Score'] = df.apply(calculate_score, axis=1)

# 3. Filter for 'Critical' 
critical_alerts = df[df['Threat_Score'] >= 8]

print(f"Analysis Complete! Found {len(critical_alerts)} critical security threats.")
critical_alerts[['Source IP Address', 'Packet Length', 'Severity Level', 'Threat_Score']].head()
```

```python
# This saves your hard work into a new CSV file
df.to_csv('/Users/beerbhadarsingh/Documents/log analysis/processed_security_data.csv', index=False)
print("File saved! You are ready for Tableau.")
```
